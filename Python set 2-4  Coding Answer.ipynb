```python
3.Write a Python function to check if a given string is a palindrome.
Expected Output (for input "madam"):
madam is a palindrome
```

```python
#Answer:
def palindrom(name):
    name=name[::-1]
    if name==name[::-1]:
        print("madam is a palindrom")
    else:
        print("madam is a palindrom")
```

```python
palindrom("madam")
```

```python
#Answer:
def find_vowel():
    name="education"
    count=0
    for i in name:
        if i in "aeiou":
            count=count+1
    print("Number of vowel:",count)        
           
```

```python
find_vowel()
```
